```python
# define base path and columns/rows of interest
base_path = "./Task2"
mycols = ['Country Name', 'Country Code']
yr = 2013
while yr < 2025:
    mycols.append(str(yr))
    yr += 1
myrows = ['WLD','NAC','EAS','ECS','LCN','SSF','SAS','MEA']
```

```python
# Deciding region and plot GDP
df = pd.read_csv(
    base_path + r"\data\GDP.csv"
)
#df.isna().sum().to_csv(base_path + "./nan_GDP.csv")
df_s = df.loc[:, mycols]
# Summing up GDP of countries in the same region
df_s = (
    df_s
    .groupby("Country Code", as_index=False)
    .sum(numeric_only=True)
)
df_s = df_s[df_s['Country Code'].isin(myrows)]
df_s = df_s.reset_index(drop=True).T  # transpose
df_s = df_s.reset_index().rename(columns={'index': 'Year'}) # reset index and rename
df_s.columns = df_s.iloc[0]                                 # set first row as header
df_s.rename(columns={'Country Code': 'Year'}, inplace=True) # rename column since transposing
df_s = df_s[1:]                                # remove first row since it is now header
df_s = df_s.apply(pd.to_numeric, errors='coerce')
years = df_s['Year'].tolist() # get list of years for x-axis
# Melt the dataframe for ggplot
df_GDP_long = df_s.melt(id_vars='Year', var_name='Region', value_name='GDP')
# color code used in the project
colors = [
    "#1f77b4", "#ff7f0e", "#2ca02c", "#d62728",
    "#9467bd", "#8c564b", "#e377c2", "#7f7f7f",
    "#bcbd22", "#17becf", "#aec7e8", "#ffbb78",
    "#98df8a", "#ff9896", "#c5b0d5", "#c49c94",
    "#f7b6d2", "#c7c7c7", "#dbdb8d", "#9edae5"
]
p =(
    ggplot(df_GDP_long, aes(x='Year', y='GDP', color='Region'))
    + geom_line()
    + geom_point()
    + scale_color_manual(values=colors[:len(df_GDP_long['Region'].unique())])
    + scale_x_continuous(breaks=years)
    + scale_y_continuous(labels=lambda l: [f"{v:.2e}" for v in l])
    + labs(title="GDP for different regions", x="Year", y="GDP")
    + theme(axis_text_x=element_text(rotation=45, ha='right'))
)
p.save(base_path + r"\img\GDP_regions.png", dpi=300, limitsize=False)
```

```python
# Combine top 3 coins data from coinmarketcap and make them as yearly time series for correlation analysis
df_b = pd.read_csv(
    r".\Task1\data\Bitcoin_monthly_010413-011125.csv",
    sep=";"
)
df_b['timeOpen'] = pd.to_datetime(df_b['timeOpen'], utc=True).dt.tz_convert(None)           
df_b = df_b.sort_values('timeOpen', ascending=True).reset_index(drop=True)

df_e = pd.read_csv(
    r".\Task1\data\Ethereum_monthly_010915-011125.csv",
    sep=";"
)
df_e['timeOpen'] = pd.to_datetime(df_e['timeOpen'], utc=True).dt.tz_convert(None)             
df_e = df_e.sort_values('timeOpen', ascending=True).reset_index(drop=True)

df_t = pd.read_csv(
    r".\Task1\data\Tether_monthly_010315-011125.csv",
    sep=";"
)
df_t['timeOpen'] = pd.to_datetime(df_t['timeOpen'], utc=True).dt.tz_convert(None)                
df_t = df_t.sort_values('timeOpen', ascending=True).reset_index(drop=True)

convert_yearly = lambda df: (
                                df
                                .set_index("timeOpen")
                                .resample("Y")["volume"]
                                .sum()
                            )
b_v = convert_yearly(df_b)
e_v = convert_yearly(df_e)
t_v = convert_yearly(df_t)

df_yearly_total = (
    b_v
    .add(e_v, fill_value=0)
    .add(t_v, fill_value=0)
    .reset_index()
)
df_yearly_total["year"] = df_yearly_total["timeOpen"].dt.year
df_yearly_total = df_yearly_total[["year", "volume"]]
df_yearly_total.set_index("year", inplace=True)
```

```python
gdp_region = {region : g['GDP'] for region,g in df_GDP_long.groupby('Region')}
inf_region = {region : g['INF'] for region,g in df_inf_long.groupby('Region')}
une_region = {region : g['UNE'] for region,g in df_une_long.groupby('Region')}
```

```python
# Function to compute and plot Two-Sided CCF Heatmap
import numpy as np
from statsmodels.tsa.stattools import ccf

lags_to_check = 2 
def ccf_heatmap_crypto_econ(df_yearly_total: pd.DataFrame, metrix_region: dict, lag_limit: int, title: str) -> pd.DataFrame: # Two-sided CCF heatmap for crypto and economic metric
    ccf_results = {} # Store CCF results for each region
    for region, metrix in metrix_region.items():
        # Align crypto and GDP on the intersection of their years
        crypto_s, metrix_aligned = df_yearly_total.align(metrix, join="inner", axis=0)
        
        #  Make Stationary (Log-Difference) conversion from raw values into "Growth Rates" (Percentage Change)
        crypto_st = np.log(crypto_s).diff().dropna()
        metrix_st = np.log(metrix_aligned).diff().dropna()
        
        # Re-align again just in case differencing created NaNs at edges
        crypto_st, metrix_st = crypto_st.align(metrix_st, join="inner", axis=0)
    

        # Compute Two-Sided CCF, adjusted=False to keep correlation between -1 & 1 and lag_limit + 1 because lag 0 is included
        # Forward: Crypto(t) vs GDP(t+k) -> "Does Crypto predict GDP?"
        forward = ccf(crypto_st, metrix_st, adjusted=False, nlags=lag_limit + 1)
        
        # Backward: GDP(t) vs Crypto(t+k) -> "Does GDP predict Crypto?"
        backward = ccf(metrix_st, crypto_st, adjusted=False, nlags=lag_limit + 1)
        
        # Combine into one array
        # Structure: [Backward(-2), Backward(-1), Lag 0, Forward(+1), Forward(+2)]
        # backward[1:][::-1] removes Lag 0 (duplicate) and reverses order
        full_ccf = np.concatenate([backward[1:][::-1], forward])
        ccf_results[region] = full_ccf

    # Create the index for the structure 
    lag_index = np.arange(-lag_limit, lag_limit + 1)

    # create DataFrame for the lagged CCF results
    ccf_df = pd.DataFrame(ccf_results, index=lag_index)
    
    # Melt for heatmap and ggplot it with number printed in the graph
    ccf_long = ccf_df.reset_index().rename(columns={"index": "lag"}).melt(id_vars="lag", var_name="Region", value_name="CCF")
    p = (
        ggplot(ccf_long, aes(x="Region", y="lag", fill="CCF"))
        + geom_tile(color="white")
        + geom_text(
            aes(label=ccf_long['CCF'].apply(lambda x: f"{x:.2f}")),
            size=8,        
            color="black"  
        )
        + scale_fill_gradient2(
            low="blue",
            mid="white",
            high="red",
            midpoint=0,
            name="CCF"
        )
        + scale_y_continuous(breaks=ccf_long['lag'].unique()) 
        + labs(
            title=title,
            x="Region",
            y="Lag"
        )
        + theme_minimal()
        + theme(
            figure_size=(10, 6)
        )
    )
    p.save(f"./Task2/img/{title.replace(' ', '_')}_Heatmap.png", dpi=300, limitsize=False)
    return ccf_df

```

```python
# Generate heatmaps for GDP, Inflation, and Unemployment against Crypto Volume
heatmap_gdp_vol = ccf_heatmap_crypto_econ(df_yearly_total, gdp_region, lags_to_check,"Cross-Correlation Function between GDP and Crypto Volume")
heatmap_inf_vol = ccf_heatmap_crypto_econ(df_yearly_total, inf_region, lags_to_check,"Cross-Correlation Function between Inflation and Crypto Volume")
heatmap_une_vol = ccf_heatmap_crypto_econ(df_yearly_total, une_region, lags_to_check,"Cross-Correlation Function between Unemployment and Crypto Volume")
```
